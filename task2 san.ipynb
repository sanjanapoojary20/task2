```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("personalized_medication_dataset.csv")
```

```python
data
```

```python
print(data.isnull().sum())
```

```python
# Fill missing categorical values with mode

data['Chronic_Conditions'].fillna(data['Chronic_Conditions'].mode()[0], inplace=True)

data['Drug_Allergies'].fillna(data['Drug_Allergies'].mode()[0], inplace=True)

data['Genetic_Disorders'].fillna(data['Genetic_Disorders'].mode()[0], inplace=True)

data['Recommended_Medication'].fillna(data['Recommended_Medication'].mode()[0], inplace=True)

data['Dosage'].fillna(data['Dosage'].mode()[0], inplace=True)

data['Duration'].fillna(data['Duration'].mode()[0], inplace=True)
```

```python
print(data.isnull().sum())
```

```python
print(data.duplicated().sum())
```

```python
#Diagnosis distribution
```

```python
data['Diagnosis'].value_counts().plot(kind='bar',color=['red','blue','green','orange','purple'])
plt.title("Diagnosis Distribution")
plt.show()
```

```python
#Age Distribution
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    data['Age'],
    bins=10,
    color='skyblue',
    edgecolor='black'
)

plt.title("Age Distribution", fontsize=16)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#Gender Ratio
```

```python
plt.figure(figsize=(6,6))

data['Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['pink','lightblue','lightgreen']
)
plt.title("Gender Ratio", fontsize=16)
plt.ylabel("")
plt.show()
```

```python
#Medication Usage
```

```python
plt.figure(figsize=(8,5))

data['Recommended_Medication'].value_counts().plot(
    kind='bar',
    color=['gold','cyan','magenta','lime','orange']
)

plt.title("Medication Usage", fontsize=16)
plt.xlabel("Medication")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
#BMI Distribution
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    data['BMI'],
    bins=10,
    color='violet',
    edgecolor='black'
)

plt.title("BMI Distribution", fontsize=16)
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.grid(linestyle='--', alpha=0.7)

plt.show()
```

```python
#Disease vs Gender
```

```python
pd.crosstab(data['Diagnosis'], data['Gender']).plot(
    kind='bar',
    figsize=(10,6),
    colormap='Set2'
)

plt.title("Disease vs Gender", fontsize=16)
plt.xlabel("Disease")
plt.ylabel("Count")
plt.show()
```

```python
#Age vs BMI
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    data['Age'],
    data['BMI'],
    color='darkblue'
)

plt.title("Age vs BMI", fontsize=16)
plt.xlabel("Age")
plt.ylabel("BMI")
plt.show()
```

```python
#Correlation Heatmap
```

```python
import seaborn as sns

plt.figure(figsize=(8,5))

sns.heatmap(
    data.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap", fontsize=16)

plt.show()
```
